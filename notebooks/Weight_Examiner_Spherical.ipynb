```python
def convert_positions(phi):
    ptsnew = np.zeros([3, np.size(phi,1)])
    XY = np.cos(phi[0,:])
    ptsnew[0,:] = np.cos(phi[0,:])*np.cos(phi[1,:])
    ptsnew[1,:] = np.cos(phi[0,:])*np.sin(phi[1,:])
    ptsnew[2,:] = np.sin(phi[0,:])
    return ptsnew
```
